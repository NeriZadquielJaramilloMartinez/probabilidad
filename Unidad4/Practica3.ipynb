```python
from scipy.stats import poisson

##Paramentro de la distribucion de poisson
media_accidentes=2

##Parte 1:Probsbilidad que no haya accidentes esta semana
probabilidad_0_accidentes=poisson.pmf(0,media_accidentes)

print("1.Probabilidad de que no haya accidentes en 1 semana: ",probabilidad_0_accidentes)

##Parte 2:Probabilidad de que q lo sumo haya tres accidentes en dos semanas
##Convertimos la media a una media para dos semanas
media_2_semanas=media_accidentes*2

##Calculamos la probabilidad acumulada hasta 3 accidetnes en dos semanas
probabilidad_3_o_menos_accidentes=poisson.pmf(3,media_2_semanas)

print("2.Probabilidad de que a lo sumo haya tres accidentes en 2 semanas: ",probabilidad_3_o_menos_accidentes)

##Ejercicio 2
##Parametros de la distribuciobn de poisson
media_muertos=5000*0.001 ##5000 hombras con una posibilidad de muertes del 0.001

##Numero de reclamaciones
x=4

##Calcular la probabilidad de que la compañia tenga que pagar 4 reclamaciones en un año determinado
probabilidad_4_reclamaciones=poisson.pmf(x,media_muertos)

print("La probabilidad de que la compañia tenga que pagar 4 reclamaciones es de: ",probabilidad_4_reclamaciones)
```
