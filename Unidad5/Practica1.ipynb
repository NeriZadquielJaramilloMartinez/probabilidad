```python
from sklearn.linear_model import LinearRegression ##Modelo de regresion lineal
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

##Datos de entrada
periodos=[2018,2019,2020,2021,2022,2023]
inversion=np.array([2,3,5,4,11,5]).reshape(-1,1)
ganancias=[20,25,34,30,40,31]

##Crear y entrenar el modelo de regresion lineal
modelo=LinearRegression()
modelo.fit(inversion,ganancias)

##Predecir la ganancias para el año 2024
ganancias_2024=modelo.predict([[8]])

print("La ganancia esperada para el 2024 es de : ",ganancias_2024[0])

##Coeficiente de regresion
print("Coeficiente de regresion de la pendiente (m):",modelo.coef_[0])
print("Interpretacion de la recta (b):",modelo.intercept_)

##Definir el tamaño de la figura
plt.figure(figsize=(6,4))
inversion=[2,3,5,4,11,5]
ganancias=[20,25,34,30,40,31]
##Graficar los datos de entrada
sns.scatterplot(x=inversion, y=ganancias)
plt.title('Diagrama de Dispercion de Inversion vs Ganacias')
plt.xlabel('Inversion(millones)')
plt.ylabel('Ganacias(millones)')
plt.grid(True)
plt.show()

##Crea en mapa de dispercion lineal de la regresion


```

```python
import matplotlib.pyplot as plt
import seaborn as sns

##Definir tamaño de la figura
plt.figure(figsize=(6,4))

##Datos de la entrada
inversion=[2,3,5,4,11,5]
ganancias=[20,25,34,30,40,31]

##Creau un diagrama de dispercion con la linea de regresion
sns.set(style="whitegrid")
sns.regplot(x=inversion,y=ganancias,scatter=True,color='blue')
plt.title('Diagrama de Dispercion con linea de regresion')
plt.xlabel('Inversion(millones)')
plt.ylabel('Ganacias(millones)')
plt.grid(True)
plt.show()

```
