```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

pmedia=70
desv_std=4
cant=500

datos=np.random.normal(pmedia,desv_std,cant)
datos[0:50]

x=np.linspace(pmedia-3*desv_std,pmedia+3*desv_std,cant)
pdf=1/(desv_std*np.sqrt(2*np.pi))*np.exp(-(x-pmedia)**2/(2*desv_std**2))
print("Imprimir los primeros 50 valores de x:\n",x[0:50])

peso_limite=68
peso_Limite=72

prob_menos_68kg=norm.cdf(peso_limite,pmedia,desv_std)
prob_menos_72kg=norm.cdf(peso_Limite,pmedia,desv_std)

prob_mas_72kg=1-prob_menos_72kg
estudiantes_menos_68kg=cant*prob_menos_68kg
estudiantes_mas_72kg=cant*prob_mas_72kg

print("Probabilidad de que un estudiante pese menos de 68kg:",prob_menos_68kg)
print("Número de estudiantes que pesan menos de 68kg:",estudiantes_menos_68kg)
print("Probabilidadad de que un  estudiante pese mas de 72 kg:",prob_mas_72kg)
print("Número aproximado de estudiantes que pesan mas de 72 kg:",estudiantes_mas_72kg)


x=np.linspace(pmedia-4*desv_std,pmedia+4*desv_std,1000)
y=norm.pdf(x,pmedia,desv_std)

plt.plot(x,y,'b',linewidth=2)
plt.fill_between(x,y,where=(x<=peso_limite),color='skyblue',alpha=0.5)
plt.title('Probabilidades de que un estudiante pese menos de 68kg. \n'r'$\mu=70$,$\sigma=4$,x=500')
plt.xlabel('Peso (KG)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
##Graficacion de la distribucion de la mas
x=np.linspace(pmedia-4*desv_std,pmedia+4*desv_std,1000)
y=norm.pdf(x,pmedia,desv_std)

plt.plot(x,y,'b',linewidth=2)
plt.fill_between(x,y,where=(x>=peso_Limite),color='skyblue',alpha=0.5)
plt.title('Probabilidades de que un estudiante pese mas de 72kg. \n'r'$\mu=70$,$\sigma=4$,x=500')
plt.xlabel('Peso (KG)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()

##Grafuica de la distribuicion normal ded menos
x=np.linspace(pmedia-4*desv_std,pmedia+4*desv_std,1000)
y=norm.pdf(x,pmedia,desv_std)

plt.plot(x,y,'b',linewidth=2)
plt.fill_between(x,y,where=(x>=peso_limite)&(x<=peso_Limite),color='skyblue',alpha=0.5)
plt.title('Probabilidades de que un estudiante pese mas de 68kg y menos de 72 kg. \n'r'$\mu=70$,$\sigma=4$,x=500')
plt.xlabel('Peso (KG)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()

```
