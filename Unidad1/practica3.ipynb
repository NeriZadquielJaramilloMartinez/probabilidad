```python
##Basado en el LIBRO estadistica para administracion pagina 12 creamos una lista vacia que representa la edad de 100 personas{
import random
edades=[]
##Hacemos un ciclo para generar 100 numeros aleatorios que representa las edades de las personas
for i in range (100):
  ##  Generamos numeros aleatorios con la funcion randint y lo guardamos cn la variable: numero_aleatorio
  numero_aleatorio=random.randint(18,70)
  ##Agregamos el numero aleatorio al la lista de edades
  edades.append(numero_aleatorio)
print(edades)
```

```python
import pandas as pd
##Crea un DataFrame de pandas con los numeros aleatorios

df=pd.DataFrame(edades,columns=['Edad'])
df.head()

```

```python
##Calcular tsblsde de distribucion de frecuencias
tabla_frecuencias=df['Edad'].value_counts().reset_index()
tabla_frecuencias.columns=['Edad','Frecuencia']
tabla_frecuencias=tabla_frecuencias.sort_values(by='Edad')
##Imprimir la tabla de distribucion de frecuencias
print("Tabla de Distribucion de Frecuencias: ")
tabla_frecuencias
```

```python
##Calcular la frecuencia relativa
tabla_frecuencias['Frec Realtiva']=tabla_frecuencias['Frecuencia']/100
tabla_frecuencias
```

```python
tabla_frecuencias['Frec  Porcentual']=tabla_frecuencias['Frecuencia']/100*100
tabla_frecuencias
```

```python
tabla_frecuencias['Frec Acumulada']=tabla_frecuencias['Frecuencia'].cumsum()
tabla_frecuencias
```

##Regla de struges
###1.Es una regla que sirve para csalcular el numero de clase idóneo en los que se debe dividir un conjunto de datos

###2.Establece que *el numero de las clases es igual a uno más el logaritmo en base dos números totaL de datos

c=1+log2(N)

donde N es el tamaño de la muestra


```python
##Importa la libreria dew maatematicas
import math
```

```python
##Volvemos a imprimir las edades
print(edades)
```

```python
##Crewa UN DataFrame con las edades
df=pd.DataFrame({'Edad':edades})
df
```

```python
##Calcule el numero de intervalos utilizando la regla de Struges
n=len(edades)
k=int(1+math.log2(n))
print("Valor de n= ",n,"\nNumeros de intervalos =",k)
```

```python
##  Crea tabla de frecuencias
tabla_frecuencia=pd.cut(df['Edad'],bins=k,include_lowest=True).value_counts().reset_index()
tabla_frecuencia.columns=['Clase','Frec']
tabla_frecuencia
```

```python
tabla_frecuencia = tabla_frecuencia.sort_values(by='Clase')
tabla_frecuencia
```

```python
##Genera columnas ,siguentes
tabla_frecuencia['Frec Rel']=tabla_frecuencia['Frec']/n
tabla_frecuencia['Frec Porc']=tabla_frecuencia['Frec']/n*100
tabla_frecuencia['Frec AC']=tabla_frecuencia['Frec'].cumsum()
tabla_frecuencia['Frec Rel Ac']=tabla_frecuencia['Frec Rel'].cumsum()
tabla_frecuencia['Frec Porc Ac']=tabla_frecuencia['Frec Porc'].cumsum()

print(tabla_frecuencia)
```
