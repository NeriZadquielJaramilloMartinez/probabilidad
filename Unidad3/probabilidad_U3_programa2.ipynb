```python
##Grafica de una distribucion de probabilidqad discreta con media,varianza y desviancion estandar

import matplotlib.pyplot as plt
import numpy as np

##Datos: valores de la variable aleatoria y sus posibilidades
x=np.array([5,6,7,8,9,10])
p=np.array([0.05,0.10,0.20,0.30,0.25,0.10])

##Calculo del valor esperado(media)
esperanza=np.sum(x*p)

##Calculo de psa esperanza del cuadrado
esperanza_cuadrada=np.sum(x**2*p)

##Calculo de la varianza y desviacion estandar
varianza=esperanza_cuadrada-esperanza**2
desviacion=np.sqrt(varianza)

##Mostrar resultados en la consola
print(f"E(x) = {esperanza:.2f}")
print(f"E(x**2) = {esperanza_cuadrada:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviacion estandar = {desviacion:.2f}")

##Crea un grafica de tipo 'steam'(tallos y hojas)
plt.stem(x,p,basefmt=" ",linefmt='b-',markerfmt='bo')
plt.title("Distribucion de probabilidad discreta\nCon Media y Desviacion Estandar")
plt.xlabel("Valores de la variable aleatoria X")
plt.ylabel("P(X)")
plt.grid(True,linestyle='--',alpha=0.6)

##Dibujar linea vertical en la media
plt.axvline(x=esperanza,color='red',linestyle='--',label=f"Media (E[X]={esperanza:.2f}")

##Dibujar el rango de desviaqcion estandar
plt.axvspan(esperanza-desviacion, esperanza+desviacion,color='red',alpha=0.2,label=f"+-1 Desv.Estandar (o={desviacion:.2f})")

##Mostrar la leyenda y limites
plt.legend()
plt.ylim(0,max(p)+0.05)
plt.xticks(x)

##Mostrar la grafica
plt.show()
```
