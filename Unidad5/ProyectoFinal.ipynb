```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np
import seaborn as sns

##Colocamos las tres tablas y las unios en una sola
df1=pd.read_csv("grupo1.csv")
df2=pd.read_csv("grupo2.csv")
df3=pd.read_csv("grupo3.csv")


##Aqui tenemos los histogramas
plt.figure(figsize=(10,5))
intervalos = range(0, 110, 10)
plt.hist(df1['Calificacion'],bins=intervalos,alpha=0.5,color='b',edgecolor='black',label='Grupo1')
plt.title('Calificaciones por estudiantes grupo 1')
plt.xlabel('Calificacion')
plt.ylabel('No. Estudiantes')
plt.legend()
plt.show()
plt.figure(figsize=(10,5))
intervalos = range(0, 110, 10)
plt.hist(df2['Calificacion'],bins=intervalos,alpha=0.5,color='b',edgecolor='black',label='Grupo2')
plt.title('Calificaciones por estudiantes grupo 2')
plt.xlabel('Calificacion')
plt.ylabel('No. Estudiantes')
plt.legend()
plt.show()
plt.figure(figsize=(10,5))
intervalos = range(0, 110, 10)
plt.hist(df3['Calificacion'],bins=intervalos,alpha=0.5,color='b',edgecolor='black',label='Grupo3')
plt.title('Calificaciones por estudiantes grupo 3')
plt.xlabel('Calificacion')
plt.ylabel('No. Estudiantes')
plt.legend()
plt.show()

df1.head()
df2.head()
df3.head()

datos = df1
nombre = "Grupo 1"

horas = np.array(datos['HorasEstudio']).reshape(-1, 1)
calificaciones = datos['Calificacion']

## 3. Entrenar el modelo de regresion lineal
modelo = LinearRegression()
modelo.fit(horas, calificaciones)

## 4. Calcular Pearson (r) y R cuadrado (confiabilidad) sin usar scipy
pearson = datos['HorasEstudio'].corr(datos['Calificacion'])
r_cuadrado = modelo.score(horas, calificaciones)

print(f"=== {nombre} ===")
print(f"Ecuacion de regresion: Calificacion = ({modelo.coef_[0]:.4f}*Horas de estudio) + {modelo.intercept_:.4f}")
print(f"Coeficiente de correlacion de Pearson (r): {pearson:.4f}")
print(f"Confiabilidad del modelo: {r_cuadrado:.4f}")

## 5. Grafica de dispersion con linea de regresion
plt.figure(figsize=(6, 4))
sns.regplot(x=datos['HorasEstudio'], y=datos['Calificacion'], line_kws={"color": "red"})
plt.title(f'Diagrama de Dispersion: {nombre}')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.tight_layout()
plt.show()



datos = df2
nombre = "Grupo 2"

horas = np.array(datos['HorasEstudio']).reshape(-1, 1)
calificaciones = datos['Calificacion']

modelo = LinearRegression()
modelo.fit(horas, calificaciones)

pearson = datos['HorasEstudio'].corr(datos['Calificacion'])
r_cuadrado = modelo.score(horas, calificaciones)

print(f"===  {nombre} ===")
print(f"Ecuacion de regresion: Calificacion = ({modelo.coef_[0]:.4f}*Horas de estudio) + {modelo.intercept_:.4f}")
print(f"Coeficiente de correlacion de Pearson (r): {pearson:.4f}")
print(f"Confiabilidad del modelo: {r_cuadrado:.4f}")

plt.figure(figsize=(6, 4))
sns.regplot(x=datos['HorasEstudio'], y=datos['Calificacion'], line_kws={"color": "red"})
plt.title(f'Diagrama de Dispersion: {nombre}')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.tight_layout()
plt.show()

datos = df3
nombre = "Grupo 3"

horas = np.array(datos['HorasEstudio']).reshape(-1, 1)
calificaciones = datos['Calificacion']

modelo = LinearRegression()
modelo.fit(horas, calificaciones)

pearson = datos['HorasEstudio'].corr(datos['Calificacion'])
r_cuadrado = modelo.score(horas, calificaciones)

print(f"=== {nombre} ===")
print(f"Ecuacion de regresion: Calificacion = ({modelo.coef_[0]:.4f}*Horas de estudio) + {modelo.intercept_:.4f}")
print(f"Coeficiente de correlacion de Pearson (r): {pearson:.4f}")
print(f"Confiabilidad del modelo: {r_cuadrado:.4f}")

plt.figure(figsize=(6, 4))
sns.regplot(x=datos['HorasEstudio'], y=datos['Calificacion'], line_kws={"color": "red"})
plt.title(f'Diagrama de Dispersion: {nombre}')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.tight_layout()
plt.show()

print(" \n\n")
print("=== Respuestas ===\n")
print("Grupo 1:")
print("En respuesta 1 del grupo 1: Que pasa es que ellos por cada cantidad hay una muy lijera disminucion de -0.1 en calificacion mientras mas estudian.")
print("En respuesta 2 del grupo 1: Pasa que la direccion es negativa muy pequeña -0.05 que realmente no tiene la fuerza como para poder hacer muchos cambios lo que ya hay si no se hace mucho.")
print("En respuesta 3 del grupo 1: No ya que estan demasida dispersas y variables como para poder ecir que realmente hay una intervancion estadistica ahy dentro.")
print("En respuesta 4 del grupo 1: Esta tabla es muy poco confible ya que tiene muy pocas posibilidades de .0033 que ocurra o que de un gran cambio a lo que ya esta. ")
print("\n")
print("Grupo 2:")
print("En respuesta 1 del grupo 2: Que pasa que por hora de estudio hay un aumento de 4.2 de calificacion por estudio.")
print("En respuesta 2 del grupo 2: Pasa que la direccion es positiva con .86 y como esta cerca del 1 eso dice que esta es fuerte.")
print("En respuesta 3 del grupo 2: Si ya que dentro de los datos lo que estudian mas sus resultados muestran estadisticamente estar menos dispersas y mas en orden por lo que si hay un cambio estadistico.")
print("En respuesta 4 del grupo 2: Esta tabla es bastante fiable por que la confiabilidad es del 0.743 eso quiere decir que es muy probable que todo ocurra como dicta. ")
print("\n")
print("Grupo 3:")
print("En respuesta 1 del grupo 3: Que pasa es que ellos por cada cantidad bastante grande de 5.28 de mejora de califoicacion por hora estudiadas.")
print("En respuesta 2 del grupo 3: Pasa que la direccion es positiva con .94 y esta es mas fuerte ya que todos se acercan mas a la linea de regresion.")
print("En respuesta 3 del grupo 3: Si aqui se puede ver directamente que las calificaciones estan muy relacionadas con la calificacion.")
print("En respuesta 4 del grupo 3: Esta tabla es la mas confiable ya que que es casi es 100% asegurado de todas las demas tablas ")
print("\n")
```
