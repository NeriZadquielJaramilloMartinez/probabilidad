```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statistics import *
import math
```

```python
auto=pd.read_csv("autos2.csv",encoding="ISO-8859-1")
```

```python
##Aqui sacaremos las estadisticas descrictivos basicas y memidas de tendencias central
auto.price.min()
mini=min(auto.price)
maxi=max(auto.price)
moda=mode(auto.price)
media=mean(auto.price)
mediana=median(auto.price)
dev=auto['price'].std()
print("El valor minimo es: ",mini)
print("El valor maximo es: ",maxi)
print("La moda es: ",moda)
print("La media es: ",media)
print("La mediana es: ",mediana)
print("La desviacion estandar es: ",dev)
auto.price.describe()
```

```python
plt.hist(auto.price)
plt.title("Histogramas de los Precios de los autos con los valores de mathplot estandar")
plt.xlabel("Precio de los autos ")
plt.ylabel("Cantidad de autos (Frecuencia)")
plt.show()
```

```python
n=len(auto.price)
k=(1+math.log2(n))
print("Valor de n=",n,"\nNumeros de intervalos=",k)
```

```python
plt.hist(auto['price'],bins=9,alpha=0.5)
plt.title('Histograma de los precios de autos basado en la regla de Struges')
plt.xlabel('Precios de los autos')
plt.ylabel('Cantidad de autos (Frecuencia)')
```

```python
##Aqui ewmpezamos las cajas con bigotes de precio(price) y de tipo de autos (body-style).
sns.set_style("whitegrid")
sns.boxplot(x="body-style",y="price",data=auto,palette="pastel")
plt.title("Distribucion de Precio por tipo de auto",fontsize=14,fontweight="bold")
plt.xlabel("Tipo de Auto",fontsize=12)
plt.ylabel("PRECIO",fontsize=12)
plt.show()

```

```python
## la segunda caja de bigote de precio(price) y de tipo de rueda (drive-wheels)
sns.set_style("whitegrid")
sns.boxplot(x="drive-wheels",y="price",data=auto,palette="pastel")
plt.title("Distribucion de Precio por tipo de rueda",fontsize=14,fontweight="bold")
plt.xlabel("Tipo de Rueda",fontsize=12)
plt.ylabel("PRECIO",fontsize=12)
plt.show()
```

```python
##En esta esta se hara la ultima diagrama de cajas con bigotes de precio (price) y ubicacion del motor (engine-location)
sns.set_style("whitegrid")
sns.boxplot(x="engine-location",y="price",data=auto,palette="pastel")
plt.title("Distribucion de Precio por ubicacion del motor",fontsize=14,fontweight="bold")
plt.xlabel("Localizacion del Motor",fontsize=12)
plt.ylabel("PRECIO",fontsize=12)
plt.show()
```
