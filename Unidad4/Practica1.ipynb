```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

##Genera datos para la distribucion

media=70
desv_std=4
num_datos=500

##Genera los datos de manera aleatoria y los ponemos en una lista (array)
datos=np.random.normal(media,desv_std,num_datos)
datos[0:50] ##Mostrar los primeros 50

##Crear un  rangoo de valores para crear una densidad
x=np.linspace(media-3*desv_std,media+3*desv_std,num_datos)
##Calcular la PDF de la distribucion normal
pdf=1/(desv_std*np.sqrt(2*np.pi))*np.exp(-(x-media)**2/(2*desv_std**2))
##Para no imprimir l¿toda la lista de valores (500), desplegamos los primeros 50
print("Imprimir los primeros 50 valores de x:\n",x[0:50],"\n y los primeros 50 pdf:\n",pdf[0:50])

##Grafricar histograma
plt.figure(figsize=(6,4))
plt.hist(datos,bins=15,alpha=0.4,color='b')##Histograma de los datos
plt.title('Distribucion de la Normal')
plt.xlabel('Peso en kg de los estudiantes')
plt.grid(True)
plt.show()

##Visualizar pdf
plt.figure(figsize=(6,4))
plt.plot(x,pdf,label='Normal:(media:{};ds:{})'.format(np.round(media,2),np.round(desv_std,2)))
plt.title('Funcion de Densidad de Probabilidad (PDF)-Distribucion Normal')
plt.xlabel('Peso en kg de los estudiantes')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(6,4))
##  Parametro de la distribucion normal
media=70 ##Media
desv_std=4 ##Desviacion estandar
total_estudiantes=500

##Variable para calculo de la probabilidad de que un estudiante que pese menos de 75 kg
peso_limite=75
peso_Limite=79

##Primero nesesitamos calcular la probabilidad de que un estudiante pese menos
##De 75 kg ur¿tiliza la funcion de distribucion acumulada (CDF) de la distribucion normal.
prob_menos_75kg=norm.cdf(peso_limite,media,desv_std)
prob_menos_79kg=norm.cdf(peso_Limite,media,desv_std)

##La probabilidad de que  un estudiante pese mas de 79 kg es complementaria
prob_mas_79kg=1-prob_menos_79kg

##Calcula del numero de estudiantes que pesan menos de 75kg
estudiantes_menos_75kg=total_estudiantes*prob_menos_75kg
estudiantes_mas_79kg=total_estudiantes*prob_mas_79kg

print("Probabilidad de que un estudiante pese menos de 75kg:",prob_menos_75kg)
print("Número de estudiantes que pesan menos de 75kg:",estudiantes_menos_75kg)
print("Probabilidadad de que u  estudiante pese mas de 79 kg:",prob_mas_79kg)
print("Número aproximado de estudiantes que pesan mas de 79 kg:",estudiantes_mas_79kg)

##Grafuica de la distribuicion normal ded menos
x=np.linspace(media-4*desv_std,media+4*desv_std,1000)
y=norm.pdf(x,media,desv_std)

plt.plot(x,y,'b',linewidth=2)
plt.fill_between(x,y,where=(x>=peso_limite)&(x<=peso_Limite),color='skyblue',alpha=0.5)
plt.title('Probabilidades de que un estudiante pese mas de 75kg y menos de 79 kg. \n$\mu=70$,$\sigma=4$,x=500')
plt.xlabel('Peso (KG)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()

##Graficacion de la distribucion de la mas
x=np.linspace(media-4*desv_std,media+4*desv_std,1000)
y=norm.pdf(x,media,desv_std)

plt.plot(x,y,'b',linewidth=2)
plt.fill_between(x,y,where=(x>=peso_Limite),color='skyblue',alpha=0.5)
plt.title('Probabilidades de que un estudiante pese mnenos de 75kg. \n$\mu=70$,$\sigma=4$,x=500')
plt.xlabel('Peso (KG)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```
