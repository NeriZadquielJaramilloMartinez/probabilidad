###Ejemplo 3 histogramas de 2 grupos

```python
#Importamos libreria para generrar numeros aleatoria
import random
```

```python
fabrica1=[]
for i in range(100):
  numero_aleatorio=round(random.uniform(50,100),2)
  fabrica1.append(numero_aleatorio)
print(fabrica1)
```

```python
fabrica2=[]
for i in range(100):
  numero_aleatorio=round(random.uniform(70,100),2)
  fabrica2.append(numero_aleatorio)
print(fabrica2)
```

```python
#Importamos una libreria nueva pars grsficar valores
import matplotlib.pyplot as plt
```

```python
plt.hist(fabrica1)
plt.title("Produccion diaria d la Fabrica 1")
```

```python
plt.hist(fabrica2)
plt.title("Produccion diaria d la Fabrica 2")
```

```python
media1=mean(fabrica1)
mediana1=median(fabrica1)
moda1=mode(fabrica1)
print("Valores de la fabrica 1")
print("La media de la fabrica 1 es=>",media1)
print("La mediana de la fabrica 1 es=>",mediana1)
print("La moda de la fabrica 1 es=>",moda1)
```

```python
media2=mean(fabrica2)
mediana2=median(fabrica2)
moda2=mode(fabrica2)
print("Valores de la fabrica 2")
print("La media de la fabrica 2 es=>",media2)
print("La mediana de la fabrica 2 es=>",mediana2)
print("La moda de la fabrica 2 es=>",moda2)
```
