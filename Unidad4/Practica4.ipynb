```python
import numpy as np
from scipy.stats import t
import matplotlib.pyplot as plt

##Datos de la muestra
data=np.array([118,125,130,128,122,135,126,132,129,120,127,127])
n=len(data)
media=np.mean(data)
desv_std=np.std(data,ddof=1)##Desviacion estandar muestral (ddof=1)
alpha=0.05 ##Nivel de significacia(95% de confianza)
df=n-1

##Valor critico de t (dos colas)
t_critico=t.ppf(1-alpha/2,df)

##Error estandar
error=t_critico*(desv_std/np.sqrt(n))

##Intervalo de confianza
lim_inf=media-error
lim_sup=media+error

print(f"Media muestral = {media:.2f} mmHg")
print(f"Desviacion estandar = {desv_std:.2f} mmHg")
print(f"Grados de libertad = {df}")
print(f"Valor critico = {t_critico:.3f}")
print(f"Intervalo de confianza 95%: [{lim_inf:.2f},{lim_sup:.2f}]mnHg")

x=np.linspace(media-3*desv_std,media+3*desv_std,200)
y=t.pdf((x-media)/(desv_std/np.sqrt(n)),df)
plt.figure(figsize=(8,5))
plt.plot(x,y,color='navy',label='Distributión t de la media muestral')
plt.axvline(lim_inf,color='red',linestyle='--',label='Limite inferior')
plt.axvline(lim_sup,color='red',linestyle='--',label='Limite superior')
plt.axvline(media,color='red',linestyle='--',label='media')
plt.title("Intervalo de confianza del 95% para la media (t de student)")
plt.xlabel("Presion arterial (mmHg)")
plt.ylabel("Densidad de probabilidad")
plt.legend()
plt.grid(True)
plt.show()
```
