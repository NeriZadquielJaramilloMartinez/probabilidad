```python
import matplotlib.pyplot as plt
import numpy as np
import math

x=np.array([0,1,2,3,4,5])
p=np.array([0.10,0.25,0.30,0.20,0.10,0.05])

vep=0
valor_esperado=0
F=np.cumsum(p)
##Verificacion de las probabilidades y funciones
for i in range(len(x)):
  xi=x[i]
  pi=p[i]
  vep+=p[i]
  producto=xi*pi
  valor_esperado+=producto


##pasos del sistema
  print(f"x= {xi}, p= {pi}, x*p={producto:.2f}")

for xi,fi in zip(x,F):
  print(f"F({xi}) = {fi:.2f}")
##Mostrar resultado final
print(f"La verificacion de la suma de las probabilidades es de: {vep:.2f} ")


##Calculo del valor esperado(media)
esperanza=np.sum(x*p)

##Calculo de psa esperanza del cuadrado
esperanza_cuadrada=np.sum(x**2*p)

##Calculo de la varianza y desviacion estandar
varianza=esperanza_cuadrada-esperanza**2
desviacion=np.sqrt(varianza)

##Mostrar resultados en la consola
print(f"E(x) = {esperanza:.2f}")
print(f"E(x**2) = {esperanza_cuadrada:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviacion estandar = {desviacion:.2f}")


##Mostrar la grafica
plt.step(x,F,where='post')
plt.title("Funcion de Distribucion aculada (FDA)")
plt.xlabel("Celulas anormales")
plt.ylabel("Probabilidad acumulada")
plt.show()

print("\nInterpretaciondes de los resultados:\n")
print(f"¿Cual es el promedio esperado de celulas anormales de la muestra: {esperanza:.2f}")
print(f"¿Existe mucha variabilidad?{varianza:.2f}")
print(f"¿Que indica la funcion acumulada sobre la posibilidad de encontrar \nun numero nbajo o alto de celulas anormales: Es de  {F[2]*100:.0f}%  entoces laas posibilidades son altas")
```
